## ✦ Exercice 1

On considère la fonction Python suivante :

```python
def suite(n):
    u = 1
    for k in range(n):
        u = (1 + 1/(k+1)) * u
    return u
```

1. Donner la forme récurrente de la suite $(u_n)$ définie par cette fonction.
2. Calculer les valeurs renvoyées par `suite(0)`, `suite(1)`, `suite(2)`.
3. Conjecturer une expression de $u_n$ en fonction de $n$.
4. Valider ou corriger la conjecture par un raisonnement **par récurrence**.

---

````python
# ⚙️ Code pour tester la fonction de l'exercice 1
def suite(n):
    u = 1
    for k in range(n):
        u = (1 + 1/(k+1)) * u
    return u

print("u_0 =", suite(0))
print("u_1 =", suite(1))
print("u_2 =", suite(2))
````

## ✦ Exercice 2

On considère la suite $(u_n)$ définie par :

$$u_0 = 2,5 \quad\text{et}\quad u_{n+1} = -u_n^{2} + 5u_n - 3$$

On donne la fonction Python suivante :

```python
def fonction(e):
    N = 0
    U = 2.5
    while abs(U - 3) >= e:
        N = N + 1
        U = -U**2 + 5*U - 3
    return N
```

1. Calculer les valeurs renvoyées par `fonction(0.1)` et `fonction(0.001)`.
2. Plus généralement, quelle est la valeur renvoyée par `fonction(e)` ?
3. Que peut-on conjecturer sur la **limite** de la suite $(u_n)$ ?

---

````python
# ⚙️ Code pour tester la fonction de l'exercice 2
def fonction(e):
    N = 0
    U = 2.5
    while abs(U - 3) >= e:
        N += 1
        U = -U**2 + 5*U - 3
    return N

print("N pour e=0.1 :", fonction(0.1))
print("N pour e=0.001 :", fonction(0.001))
````

## ✦ Exercice 3

Une ferme marine produit des algues alimentaires. Un expert estime que la masse totale du stock peut être modélisée par la suite $(u_n)$ :

$$u_n = \frac{5\,000}{9 + e^{-0.1n}}$$

où $n$ est exprimé en **semaines**.

1. 
   1. À l’aide d’un outil numérique, calculer les 10 premiers termes de $(u_n)$.
   2. Conjecturer les variations et la limite de $(u_n)$.
   3. Valider ou corriger ces conjectures.
   4. Interpréter le résultat dans le contexte.

2. Le gérant souhaite savoir à partir de quelle semaine le stock dépassera **550 tonnes**.

   1. Compléter la fonction Python ci-dessous pour obtenir la liste des stocks hebdomadaires jusqu’à ce qu’ils dépassent `s`.
   2. Que renvoie `sum(production(550))` ? Interpréter ce résultat.

---

````python
# ⚙️ Calcul des 10 premiers termes de u_n
import math

def u(n):
    return 5000 / (9 + math.exp(-0.1 * n))

vals = [u(n) for n in range(10)]
vals
````

````python
# ⚙️ Fonction à compléter pour le suivi du stock jusqu'à ce qu'il dépasse s
def production(s):
    p = [u(0)]
    n = 0
    while p[-1] <= s:
        n += 1
        p.append(u(n))
    return p

stocks = production(550)
print("Nombre de semaines jusqu'à dépasser 550 t :", len(stocks) - 1)
print("Somme des stocks sur cette période :", sum(stocks))
````
